```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

features = np.array([
    [1, 4, 1.5, 3.8],
    [3, 6, 1.0, 3.1],
    [4, 7, 0.8, 3.9],
    [6, 8, 0.2, 2.8],
    [7, 9, 0.2, 2.8]
])

sal=np.array([45,60,72,85,95])

sns.scatterplot(x=features[:, 3:].ravel(), y=sal)
plt.xlabel('CGPA')
plt.ylabel('salary')
plt.grid()
```

```python
def make_prediction(x,y,w,b):
  m=x.shape[0]
  prediction_list=np.zeros(m)
  for i in range(m):
    prediction_list[i]=np.dot(x[i],w)+b
  return prediction_list
```

```python
m= features.shape[0]
n=features.shape[1]

w_init=np.ones((n,))
b_init=1.0

make_prediction(features,sal,w_init,b_init)
```

```python
def compute_cost(x, y, w, b):
    m = x.shape[0]
    cost = 0.0

    predictions = make_prediction(x, y, w, b)
    print('predictions :', predictions)
    errors= predictions - y
    print('errors :', errors)
    cost = np.sum(errors ** 2)

    # for i in range(m):
    #     prediction = w * x[i] + b
    #     error = prediction - y[i]
    #     cost += error ** 2

    cost = cost / (2 * m)
    return cost
```

```python
w_init=np.ones((n,))
b_init=1.0

compute_cost(features,sal,w_init,b_init)
```
